```python
data_folder = "../data"
```

```python
import pandas as pd

df=pd.read_csv(data_folder + '/chembl_train.csv')
```

```python
import matplotlib.pyplot as plt
_=plt.hist(df.length,bins=50)
```

```python
tokens = {}
for mol in df.selfies[:100000]:
    ts=mol[1:-1].split("][")
    for t in ts:
        if t in tokens:
            tokens[t]+=1
        else:
            tokens[t]=1
```
